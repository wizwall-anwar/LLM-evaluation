## 8️⃣ View Results

```python
import pandas as pd
import json
from pathlib import Path

# Load and display results
results_dir = Path('results')

all_results = []
for result_file in results_dir.rglob('*.json'):
    with open(result_file) as f:
        data = json.load(f)
        
    meta = data.get('metadata', {})
    metrics = data.get('aggregate_metrics', {})
    
    all_results.append({
        'Model': meta.get('model'),
        'Prompt': meta.get('prompt_template'),
        'Dataset': meta.get('dataset'),
        'EM': metrics.get('overall', {}).get('em', 0),
        'F1': metrics.get('overall', {}).get('f1', 0),
        'Ans_Accuracy': metrics.get('answerability_detection', {}).get('accuracy', 0),
        'Avg_Time': metrics.get('performance', {}).get('avg_response_time', 0),
        'Tokens': metrics.get('performance', {}).get('total_tokens', 0)
    })

# Create DataFrame
df = pd.DataFrame(all_results)
print("\n" + "="*80)
print("EVALUATION RESULTS SUMMARY")
print("="*80)
print(df.to_string(index=False))

# Display as styled table
display(df.style.background_gradient(subset=['EM', 'F1'], cmap='RdYlGn'))
```

## 9️⃣ Display Visualizations

```python
from IPython.display import Image, display
import os

# Find all visualization directories
viz_dirs = list(Path('results').rglob('visualizations'))

for viz_dir in viz_dirs:
    dataset_name = viz_dir.parent.name
    print(f"\n{'='*80}")
    print(f"VISUALIZATIONS - {dataset_name.upper()}")
    print('='*80)
    
    # Display each visualization
    viz_files = ['performance_comparison.png', 
                 'answerability_confusion_matrix.png',
                 'performance_metrics.png',
                 'comparison_table.png']
    
    for viz_file in viz_files:
        img_path = viz_dir / viz_file
        if img_path.exists():
            print(f"\n{viz_file}:")
            display(Image(filename=str(img_path)))

print("\n✅ All visualizations displayed!")
```
